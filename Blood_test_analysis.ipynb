```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI 

load_dotenv()

llm=ChatGoogleGenerativeAI(model="gemma-4-31b-it",google_api_key="Enter your api key ")
```

```python
with open("Blood_test_report.txt","r") as f:
    blood_test=f.read() 


print(blood_test[:200])
```

```python
extraction_prompt=f"""
Yor are a model data extraction assistant.

From the blood test report below,extract All test values and classify each one as HIGH,LOW or NORMAL
based on the reference ranges provided in report 

Format your response as :
- Test Name: value | Status: HIGH/LOW/NORMAL | Reference: range 

Blood Report:
{blood_test}
""" 

extraction_response=llm.invoke(extraction_prompt)

extraction_response=extraction_response.text
print("=== STAGE 1: EXTRACTION VALUES ===")
print(extraction_response)



```

# Suggesting the Diet PLan According to Test Report 

```python
diet_prompt=f"""
You a clinical nutritionist specializing in Indian dietry habits.
Based on the blood work analysis below, write:
1. A short health summary in 4-5 lines explaning the patient's condition in simple language
2. A short ,practical Indian diet plan having only twop section (1) Foods to avoid (2) Foods to eat more of

Blood work Analysis
{extraction_response}
"""

diet_reponse=llm.invoke(extraction_response)
diet_reponse=diet_reponse.text

print("=== STAGE 2: HEALTH SUMMARY & DIET PLAN ===")
print(diet_reponse)
```
